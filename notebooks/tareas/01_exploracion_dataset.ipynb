# ¿Qué datos genera?
Netflix recopila datos de comportamiento en tiempo real: qué series ve cada usuario, cuándo pausa, rebobina o abandona, qué busca, qué valoraciones otorga y desde qué dispositivo se conecta. Se generan millones de eventos por segundo, alcanzando varios terabytes diarios. Los datos incluyen metadatos de contenido, historial de visualización, horarios de uso y ubicación geográfica.

# ¿Quién los usa?
Los datos son consumidos por equipos internos de ingeniería y ciencia de datos para entrenar algoritmos de ML. Los productores y ejecutivos de contenido los usan para decisiones de inversión en series originales. Los socios de distribución (Smart TV, consolas) acceden a métricas de uso. Los anunciantes del plan con anuncios reciben datos agregados de audiencia.

# ¿Qué problema resuelve?
Resuelve el problema de la elección y la retención de usuarios: con más de 15,000 títulos disponibles, sin recomendaciones el usuario no sabe qué ver y abandona la plataforma. El algoritmo reduce el tiempo de decisión y aumenta el tiempo de visualización. Netflix estima que el motor de recomendación genera $1,000 millones anuales en valor de retención, evitando cancelaciones de suscripción.

# ¿Por qué no bastaría Excel?
Excel colapsa ante este caso por cuatro razones: el Volumen supera los millones de filas manejables; la Velocidad requiere procesamiento en tiempo real que Excel no admite; la Variedad incluye video, audio y texto estructurado simultáneo; y la Complejidad del modelo de ML con 2 billones de parámetros es imposible en una hoja de cálculo.

# Justificación: ¿Por qué se considera Big Data?
El caso de Netflix cumple con las 4 V clásicas que definen el Big Data:
VDimensiónDescripciónVVolumenMás de 2 TB de datos de comportamiento se generan cada día. El histórico acumulado de millones de usuarios durante años representa petabytes de información que ninguna herramienta convencional puede almacenar ni procesar.VVelocidadLos datos se procesan en tiempo real: cuando el usuario pausa, retrocede o busca, el sistema actualiza instantáneamente las recomendaciones. Se necesita procesamiento de stream continuo (Apache Kafka, Flink) para reaccionar en milisegundos.VVariedadLos datos son heterogéneos: eventos de clics (estructurados), streams de vídeo, texto libre de búsquedas, valoraciones numéricas, datos de sensores de dispositivos y metadatos de contenido. No pueden unificarse en una sola tabla relacional.VVeracidadNetflix implementa sistemas robustos de validación y limpieza para garantizar que los datos de comportamiento sean fiables, eliminando eventos duplicados, erróneos o generados por bots, lo cual es crítico para la precisión del algoritmo.
