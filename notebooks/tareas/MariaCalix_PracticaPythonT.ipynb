```python
#EJERCICIO 17 Análisis de ventas con pandas
import pandas as pd

data_ventas = {
    'Fecha':         ['2024-01-05','2024-01-07','2024-01-10','2024-01-12','2024-01-15',
                      '2024-01-18','2024-01-20','2024-01-22','2024-01-25','2024-01-28',
                      '2024-02-01','2024-02-03','2024-02-06','2024-02-09','2024-02-12'],
    'Vendedor':      ['Pedro','Ana','Pedro','Luis','Ana',
                      'Luis','Pedro','Ana','Luis','Pedro',
                      'Ana','Luis','Pedro','Ana','Luis'],
    'Producto':      ['Laptop','Mouse','Teclado','Monitor','Laptop',
                      'Mouse','Monitor','Teclado','Laptop','Mouse',
                      'Laptop','Teclado','Monitor','Mouse','Laptop'],
    'Categoría':     ['Electrónica','Accesorios','Accesorios','Electrónica','Electrónica',
                      'Accesorios','Electrónica','Accesorios','Electrónica','Accesorios',
                      'Electrónica','Accesorios','Electrónica','Accesorios','Electrónica'],
    'Cantidad':      [2, 5, 3, 1, 3, 8, 2, 4, 1, 10, 2, 6, 3, 7, 4],
    'PrecioUnitario':[15000, 350, 800, 8500, 15000, 350, 8500, 800, 15000, 350,
                      15000, 800, 8500, 350, 15000]
}

df_ventas = pd.DataFrame(data_ventas)
df_ventas['TotalVenta'] = df_ventas['Cantidad'] * df_ventas['PrecioUnitario']

print("=== DataFrame con TotalVenta ===")
print(df_ventas.head())

print("\n=== Total vendido por vendedor ===")
print(df_ventas.groupby('Vendedor')['TotalVenta'].sum().sort_values(ascending=False))

por_producto = df_ventas.groupby('Producto')['Cantidad'].sum()
print(f"\nProducto más vendido: {por_producto.idxmax()} ({por_producto.max()} unidades)")

por_categoria = df_ventas.groupby('Categoría')['TotalVenta'].sum()
print(f"\nCategoría con mayor ingreso: {por_categoria.idxmax()} (L. {por_categoria.max():,.2f})")

print("\n=== Ventas ordenadas de mayor a menor ===")
print(df_ventas.sort_values('TotalVenta', ascending=False).reset_index(drop=True))
```

```python
#EJERCICIO 19 Análisis de préstamos
import pandas as pd

data_prestamos = {
    'Cliente':         ['Ana Méndez',   'Carlos Fuentes','Beatriz Ramos','David Solís',
                        'Elena Castro', 'Fernando Pinto','Gloria Medina','Hernán Rojas',
                        'Isabel Vega',  'Jorge Campos',  'Karla Bravo',  'Leonardo Paz'],
    'MontoSolicitado': [30000, 75000, 50000, 120000, 45000, 80000,
                        25000, 60000,  95000,  40000, 55000, 110000],
    'Plazo_meses':     [12, 36, 24, 60, 18, 48, 12, 24, 60, 18, 36, 48],
    'IngresoMensual':  [8000, 22000, 15000, 35000, 12000, 25000,
                        7000, 18000, 30000, 11000, 17000, 40000],
    'Estado':          ['Aprobado','Rechazado','Aprobado','Aprobado','Pendiente',
                        'Aprobado','Rechazado','Pendiente','Aprobado','Rechazado',
                        'Aprobado','Pendiente']
}

df_prest = pd.DataFrame(data_prestamos)

print(f"Monto promedio solicitado: L. {df_prest['MontoSolicitado'].mean():,.2f}")

print("\n=== Solicitudes por estado ===")
print(df_prest['Estado'].value_counts())

print("\n=== Préstamos aprobados ===")
print(df_prest[df_prest['Estado'] == 'Aprobado'].reset_index(drop=True))

print("\n=== Clientes con monto > L. 50,000 ===")
print(df_prest[df_prest['MontoSolicitado'] > 50000].reset_index(drop=True))

print("\n=== Promedio de monto por estado ===")
print(df_prest.groupby('Estado')['MontoSolicitado'].mean().round(2))
```

```python
#EJERCICIO 20 Mini análisis de datos tipo empresa
import pandas as pd

data_empleados = {
    'Nombre':       ['Laura Padilla',  'Ricardo Núñez', 'Alejandra Soto', 'Marcos Ávila',
                     'Natalia Rivas',  'Emilio Vargas', 'Daniela Espino', 'Héctor Romero',
                     'Paola Guzmán',   'Cristian Blanco','Verónica Dimas','Ramiro Castillo'],
    'Departamento': ['TI','Finanzas','RRHH','TI','Ventas','Finanzas',
                     'TI','Ventas','RRHH','TI','Ventas','Finanzas'],
    'Cargo':        ['Desarrollador','Analista','Coordinadora','Arquitecto',
                     'Ejecutiva','Contador','DevOps','Asesor',
                     'Reclutadora','QA Engineer','Supervisora','CFO'],
    'Salario':      [35000, 42000, 28000, 65000, 31000, 40000,
                     38000, 29000, 27000, 33000, 34000, 75000],
    'AñosExp':      [3, 7, 2, 12, 4, 9, 1, 6, 3, 5, 8, 15],
    'Ciudad':       ['Tegucigalpa','San Pedro Sula','Tegucigalpa','San Pedro Sula',
                     'La Ceiba','Tegucigalpa','San Pedro Sula','La Ceiba',
                     'Tegucigalpa','Comayagua','Tegucigalpa','San Pedro Sula']
}

df_emp = pd.DataFrame(data_empleados)

print("=== Primeros 5 registros ===")
print(df_emp.head())

print("\n=== Estadísticas generales de salarios ===")
print(df_emp['Salario'].describe())

print("\n=== Salario promedio por departamento ===")
print(df_emp.groupby('Departamento')['Salario'].mean().round(2).sort_values(ascending=False))

print("\n=== Empleado con salario más alto ===")
print(df_emp.loc[df_emp['Salario'].idxmax()])

print("\n=== Empleados con más de 5 años de experiencia ===")
print(df_emp[df_emp['AñosExp'] > 5].reset_index(drop=True))

print("\n=== Empleados ordenados por salario ===")
print(df_emp.sort_values('Salario', ascending=False).reset_index(drop=True))

def clasificar_experiencia(años):
    if años <= 2:
        return 'Junior'
    elif años <= 6:
        return 'Intermedio'
    else:
        return 'Senior'

df_emp['NivelExperiencia'] = df_emp['AñosExp'].apply(clasificar_experiencia)

print("\n=== DataFrame con NivelExperiencia ===")
print(df_emp[['Nombre','Departamento','AñosExp','NivelExperiencia']])

print("\n=== Distribución de niveles ===")
print(df_emp['NivelExperiencia'].value_counts())
```
